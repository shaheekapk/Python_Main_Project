<h4 style=>Before performing imputation, the missing values were reviewed based on their data types. There are 3 numerical columns and 4 text columns with missing values. <h3 style=>Step 1:Checking Skewness</h3>

<div style="border:1px solid #999; padding:10px; border-radius:5px;">
<h4 style=>Understanding Skewness</h4>

<h4 style=>Skewness is a measure that shows how much a numerical data distribution differs from a symmetric distribution. A skewness value close to zero indicates that the data is approximately symmetric, while a positive value indicates right-skewness and a negative value indicates left-skewness. Skewness is checked before imputing missing numerical values because it helps in selecting a suitable imputation method. For approximately symmetric data, the mean can be used, whereas for highly skewed data, the median is generally preferred because it is less affected by extreme values.</h4>
</div>

````python
#Checking Skewness.

df[['previous_gpa', 'attendance_percentage', 'time_management_score']].skew()
````

<h4 style=>Result_Explanation</h4>

<h4 style=>
The skewness results for previous_gpa, attendance_percentage, and time_management_score are -0.023, -0.165, and -0.355, respectively. These values are relatively close to zero, indicating that the three variables are approximately symmetric, with only slight negative skewness. Therefore, mean imputation is chosen to fill the missing values. The mean is appropriate because the distributions are not strongly skewed, so the average value can reasonably represent the missing observations without being heavily affected by extreme values.
    The skewness of the numerical variables was checked to understand the distribution of the data and to select an appropriate method for handling missing values..</h4>

<h3 style=>Step 2:Imputing Numeric Columns with Mean</h3>

````python
#Fillinhg missing values in numeric columns with mean.

df['previous_gpa'] = df['previous_gpa'].fillna(df['previous_gpa'].mean())

df['attendance_percentage'] = df['attendance_percentage'].fillna(df['attendance_percentage'].mean())

df['time_management_score'] = df['time_management_score'].fillna(df['time_management_score'].mean())
````

````python
#Varification of imputation

df[['previous_gpa', 'attendance_percentage', 'time_management_score']].isna().sum()
````

<h4 style=>Mean Imputation and Verification</h4>

<h4 style=>Step The missing values in the three numerical columns, previous_gpa, attendance_percentage, and time_management_score, were filled using mean imputation. The standard "fillna()" method was used to replace the missing values with the mean of their respective columns. After imputation, the "isnull().sum()" function was used to verify the result. All three columns returned 0 missing values, confirming that the missing numerical values were successfully imputed.</h4>


<h3 style=>Step 3:Imputing with .bfill() and .ffill()

````python
#Imputing Text Columns parent_education,notes quality and device_availabilty by using .bfill() and .ffill()

text_cols = ['notes_quality', 'sleep_quality', 'device_availability']

for col in text_cols:
    df[col] = df[col].ffill().bfill()
````

<h4 style=>Why We Use "ffill()" and "bfill()"</h4>

<h4 style=>"ffill()" and "bfill()" are used to fill missing categorical values while maintaining the existing flow of the data. "ffill()" uses the previous available value, while "bfill()" uses the next available value. Using these methods can help preserve the continuity of nearby observations and avoid leaving gaps that may affect data visualization. Since the values are filled based on the existing sequence of the data, they can maintain the overall pattern without unnecessarily introducing unrelated categories or interrupting the visual trend.<?


<h3 style=>Step 4: Imputing with "MODE"

````python
#Imputing "parent_education" Column with mode.

df['parent_education'] = df['parent_education'].fillna(df['parent_education'].mode()[0])
````

<h4 style=> Mode Imputation for Parent Education

<h4 style=>The "mode" was used to fill the missing values in the "parent_education" column because it is a categorical variable. The mode represents the most frequently occurring education category in the dataset, so it provides a reasonable representation of the common educational background of parents. Using the mode helps maintain the existing distribution of the data and avoids introducing a less common category that may not represent the overall social and educational conditions of the parents in the dataset.


````python
#Verifying Missing Values

df.isna().sum()
````

````python
# Adjusting Total Hours to 24 Hours by creating mask .loc ,div and mul were used for further steps. 

cols = [
    'study_hours_per_day',
    'physical_activity_hours',
    'sleep_hours'
]

# Calculate the original total hours
total_hours = df[cols].sum(axis=1).round(2)

# Identify records exceeding 24 hours
mask = total_hours > 24

# Adjust proportionally and round to 2 decimal places
df.loc[mask, cols] = (
    df.loc[mask, cols]
    .div(total_hours[mask], axis=0)
    .mul(24)
    .round(2)
)

# Recalculate the total hours
total_hours_after = df[cols].sum(axis=1)

# Correct any remaining totals above 24 due to rounding
mask_after = total_hours_after > 24

df.loc[mask_after, 'sleep_hours'] = (
    df.loc[mask_after, 'sleep_hours']
    - (total_hours_after[mask_after] - 24)
).round(2)
````

````python
# Line Chart with Subplots Showing Average Stress, Exam Anxiety, and Sleep Hours by Exam Score Range

import pandas as pd
import matplotlib.pyplot as plt

# Create exam score ranges as a separate variable
score_range = pd.cut(
    df['exam_score'],
    bins=[0, 20, 40, 60, 80, 100],
    labels=['0-20', '21-40', '41-60', '61-80', '81-100'],
    include_lowest=True
)

# Calculate average stress, anxiety, and sleep for each score range
grouped_data = df.groupby(score_range, observed=False).agg(
    average_stress=('stress_level', 'mean'),
    average_anxiety=('exam_anxiety_level', 'mean'),
    average_sleep=('sleep_hours', 'mean')
)

# Create the figure
fig = plt.figure(figsize=(15, 8))

gs = fig.add_gridspec(
    3, 2,
    width_ratios=[2, 1],
    height_ratios=[1, 1, 1],
    wspace=0.35,
    hspace=0.55
)

# Main chart on the left
ax1 = fig.add_subplot(gs[:, 0])

ax1.plot(
    grouped_data.index.astype(str),
    grouped_data['average_stress'],
    marker='o',
    label='Average Stress Level'
)

ax1.plot(
    grouped_data.index.astype(str),
    grouped_data['average_anxiety'],
    marker='o',
    label='Average Exam Anxiety'
)

ax1.plot(
    grouped_data.index.astype(str),
    grouped_data['average_sleep'],
    marker='o',
    label='Average Sleep Hours'
)

ax1.set_title('Stress, Anxiety, and Sleep Hours vs Exam Scores')
ax1.set_xlabel('Exam Score Range')
ax1.set_ylabel('Average Value')
ax1.legend()
ax1.grid(True, alpha=0.3)

# Small chart 1: Average Stress
ax2 = fig.add_subplot(gs[0, 1])

ax2.plot(
    grouped_data.index.astype(str),
    grouped_data['average_stress'],
    marker='o'
)

ax2.set_title('Average Stress Level', fontsize=10)
ax2.set_xlabel('Exam Score Range', fontsize=8)
ax2.set_ylabel('Stress', fontsize=8)
ax2.tick_params(axis='both', labelsize=8)
ax2.grid(True, alpha=0.3)

# Small chart 2: Average Exam Anxiety
ax3 = fig.add_subplot(gs[1, 1])

ax3.plot(
    grouped_data.index.astype(str),
    grouped_data['average_anxiety'],
    marker='o'
)

ax3.set_title('Average Exam Anxiety', fontsize=10)
ax3.set_xlabel('Exam Score Range', fontsize=8)
ax3.set_ylabel('Anxiety', fontsize=8)
ax3.tick_params(axis='both', labelsize=8)
ax3.grid(True, alpha=0.3)

# Small chart 3: Average Sleep Hours
ax4 = fig.add_subplot(gs[2, 1])

ax4.plot(
    grouped_data.index.astype(str),
    grouped_data['average_sleep'],
    marker='o'
)

ax4.set_title('Average Sleep Hours', fontsize=10)
ax4.set_xlabel('Exam Score Range', fontsize=8)
ax4.set_ylabel('Sleep Hours', fontsize=8)
ax4.tick_params(axis='both', labelsize=8)
ax4.grid(True, alpha=0.3)

# Main title and layout
fig.suptitle(
    'Student Wellbeing and Exam Performance',
    fontsize=14
)

fig.subplots_adjust(
    top=0.90,
    bottom=0.08,
    left=0.06,
    right=0.97
)

plt.show()
````

<h3 style="color:#2F5597; font-size:22px; font-weight:bold; border-bottom:2px solid #2F5597; padding-bottom:5px;">Chart 12: Relationships Between Student Factors and Exam Scores (Heatmap with Scatter Plot Subplots)</h3>



````python
#Checking relationship between factors and exam scores by using heatmap,supported by 
#scatter charts as sub charts.

import matplotlib.pyplot as plt
import seaborn as sns

# Columns for the main heatmap
main_cols = [
    'exam_score',
    'previous_exam_score',
    'attendance_percentage',
    'study_hours_per_day',
    'self_study_hours',
    'stress_level',
    'exam_anxiety_level',
    'sleep_hours'
]

fig = plt.figure(figsize=(20, 10))

gs = fig.add_gridspec(
    3, 2,
    width_ratios=[2.5, 1],
    hspace=0.6,
    wspace=0.4
)

# Main heatmap on the left
ax1 = fig.add_subplot(gs[:, 0])

sns.heatmap(
    df[main_cols].corr(),
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    vmin=-1,
    vmax=1,
    annot_kws={'size': 9},
    ax=ax1
)

ax1.set_title('Overall Correlation Heatmap', fontsize=14)
ax1.set_xticklabels(
    ax1.get_xticklabels(), rotation=45, ha='right', fontsize=9
)
ax1.set_yticklabels(
    ax1.get_yticklabels(), rotation=0, fontsize=9
)

# Small scatter plot 1: Previous vs Current Exam Score
ax2 = fig.add_subplot(gs[0, 1])

sns.scatterplot(
    data=df,
    x='previous_exam_score',
    y='exam_score',
    alpha=0.3,
    s=10,
    ax=ax2
)

ax2.set_title('Previous vs Current Exam Score', fontsize=10)
ax2.set_xlabel('Previous Exam Score')
ax2.set_ylabel('Current Exam Score')

# Small scatter plot 2: Study Hours vs Exam Score
ax3 = fig.add_subplot(gs[1, 1])

sns.scatterplot(
    data=df,
    x='study_hours_per_day',
    y='exam_score',
    alpha=0.3,
    s=10,
    ax=ax3
)

ax3.set_title('Daily Study Hours vs Exam Score', fontsize=10)
ax3.set_xlabel('Daily Study Hours')
ax3.set_ylabel('Current Exam Score')

# Small scatter plot 3: Sleep Hours vs Exam Score
ax4 = fig.add_subplot(gs[2, 1])

sns.scatterplot(
    data=df,
    x='sleep_hours',
    y='exam_score',
    alpha=0.3,
    s=10,
    ax=ax4
)

ax4.set_title('Sleep Hours vs Exam Score', fontsize=10)
ax4.set_xlabel('Sleep Hours')
ax4.set_ylabel('Current Exam Score')

fig.suptitle(
    'Correlation and Scatter Plot Analysis of Student Performance',
    fontsize=16
)

fig.subplots_adjust(
    top=0.90, bottom=0.12, left=0.08, right=0.97
)

plt.show()
````

When the entered ID matches, df['exam_score'].iloc[i] retrieves that student's exam marks, while df['pass_status'].iloc[i] retrieves their Pass/Fail result. This finding is worth investigating further because it may differ from common assumptions about educational opportunities. The data alone does not explain why this pattern occurred, and other student characteristics or circumstances may be relevant.</h4>

<h4>These observations should be used to encourage awareness of differences in opportunities, not to stereotype students or families. <h4>Based on the findings, students can be encouraged to establish realistic study schedules, attend classes regularly, practise self-study, review previous lessons, and manage their preparation time effectively. They should also be encouraged to maintain adequate sleep, take appropriate breaks, and seek clarification when they do not understand a topic. These practices can support more consistent learning, although no single routine will work equally well for every student.</h4>

<h4>Parents and teachers should consider more than examination marks when assessing a student's progress. <h4>Educational institutions can use performance data to identify broad patterns and determine where additional support may be helpful. For example, they may review attendance, previous performance, or changes in results to identify areas that deserve further investigation. <h4>Awareness is also important. Students should understand that setbacks and lower marks do not define their intelligence or future potential. Parents and teachers should recognize progress, encourage consistent effort, and provide constructive feedback. Students who experience persistent anxiety or emotional difficulties should have access to appropriate support, including qualified professionals when necessary.</h4>

<h4>The overall goal should be to create a supportive learning environment in which students can identify their difficulties, receive suitable guidance, and work towards improvement. Rather than expecting every student to achieve the same results or learn at the same pace, support should reflect individual needs, circumstances, and opportunities.</h4>

These findings provide useful areas for further investigation, but they do not prove that any one factor directly causes higher or lower marks.</h4>
